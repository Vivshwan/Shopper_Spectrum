# **Project Summary -**

# Project Overview
Shopper Spectrum is an end-to-end machine learning project that delivers actionable customer intelligence through RFM (Recency, Frequency, Monetary) analysis, K-Means clustering, and collaborative filtering recommendations. The system processes the UCI Online Retail dataset (541,909 transactions) to segment customers into four distinct behavioral groups and provide personalized product recommendations, all accessible through an interactive Streamlit web application.

# Technical Architecture

The project follows a comprehensive data science pipeline:

# Data Preprocessing

Loaded 541,909 transactions from online_retail.csv

Cleaned data by removing canceled invoices, negative quantities, and missing CustomerIDs

Calculated TotalAmount per transaction (Quantity × UnitPrice)

Reduced dataset to 392,732 valid transactions with 4,374 unique customers

# RFM Analysis

Recency: Days since last purchase (snapshot date: max invoice date + 1 day)

Frequency: Number of unique invoices per customer

Monetary: Total spending per customer

Applied StandardScaler for feature normalization

# Customer Segmentation

Implemented K-Means clustering with forced 4 clusters to create business-relevant segments

Used Elbow method, Silhouette Score, Calinski-Harabasz, and Davies-Bouldin metrics for evaluation

Achieved Silhouette Score of 0.52, indicating good cluster separation

Created custom labeling logic mapping RFM characteristics to 4 segments:

High-Value 👑: Recent, frequent, high-spending customers (best customers)

Regular ⭐: Steady purchasers with consistent behavior

Occasional 🔄: Infrequent buyers with low spending

At-Risk ⚠️: Customers with long purchase gaps needing re-engagement

# Memory Optimization

Sampled 2,000 records for Silhouette score calculation to handle large datasets

Implemented efficient data structures to minimize memory footprint

Managed large similarity matrices (1,800+ products × 1,800+ products)

# Product Recommendation System

Built collaborative filtering engine using cosine similarity

Created user-item matrix (4,374 customers × 1,880 products)

Generated product similarity matrix for "customers who bought X also bought Y" recommendations

Handles partial matches and provides fallback popular product suggestions

# Model Persistence

Saved 6 pickle files:

1. rfm_scaler.pkl: StandardScaler for consistent preprocessing

2. kmeans_final_model.pkl: Trained K-Means with 4 clusters

3. cluster_labels.pkl: Cluster ID to segment name mapping

4. cluster_centroids.pkl: Cluster center coordinates

5. streamlit_model_package.pkl: Complete package for web deployment

6. recommendation_model.pkl: Product similarity matrix

# Visualization Suite

Generated 4 comprehensive PNG visualizations:

1. cluster_metrics_comparison.png: Elbow, Silhouette, Calinski-Harabasz, Davies-Bouldin plots

2. cluster_visualizations_complete.png: PCA 2D/3D scatter plots, profiles, segment distribution

3. eda_visualizations.png: 9 subplots showing sales trends, product performance, country analysis

# Streamlit Web Application

The web app provides three core functionalities:

1. Customer Segmentation Page:

        Users input RFM values (Recency, Frequency, Monetary)

        Predicts customer segment in real-time

2. Product Recommendations Page:

# **Problem Statement**


# Business Challenge

In today's competitive e-commerce landscape, businesses collect vast amounts of transactional data but often fail to extract actionable customer insights. Organizations struggle to understand who their customers are, what drives purchasing behavior, and how to effectively engage different customer segments. Without proper customer intelligence, businesses waste marketing budgets on generic campaigns, lose high-value customers to competitors, and miss opportunities to re-engage dormant customers.

# Core Problems
1. Customer Understanding Gap

      E-commerce businesses have thousands of customers but cannot differentiate between casual browsers and loyal purchasers. Without segmentation, marketing teams treat all customers identically, resulting in ineffective communication and poor resource allocation. Managers lack clear visibility into customer lifetime value, purchase patterns, and churn risk indicators.

2. Manual Segmentation Limitations

      Traditional methods rely on demographic data (age, location, income) which provides incomplete behavioral insights. Manual segmentation using spreadsheet analysis is time-consuming, error-prone, and fails to capture complex purchasing patterns. Human-driven segmentation cannot scale with growing customer bases and misses subtle behavioral differences.

3. Personalization Deficit
      
      Without understanding customer preferences, businesses cannot offer relevant product recommendations. Generic product suggestions frustrate customers, leading to lower conversion rates and reduced basket sizes. Missing the "customers who bought X also bought Y" opportunity results in significant lost revenue.

4. Reacting Instead of Proacting
      
      Organizations only identify at-risk customers after they've stopped purchasing, leaving no time for intervention. Businesses spend disproportionate marketing budgets acquiring new customers while failing to retain existing ones. The absence of early warning signals prevents proactive retention strategies.

5. Data Underutilization
      
      While retail datasets contain rich transactional history, organizations lack the technical expertise to transform raw data into strategic insights. Machine learning models remain untapped resources that could provide competitive advantage. The gap between available data and actionable intelligence remains unresolved.

# Project Objectives

To address these challenges, we need a comprehensive solution that:

      Segments customers into distinct behavioral groups using RFM analysis

      Provides personalized product recommendations through collaborative filtering

      Delivers insights through an accessible web interface

      Identifies at-risk customers for proactive retention

      Enables data-driven marketing strategies

The solution must process large datasets efficiently, generate interpretable results, and democratize customer analytics across the organization. By implementing machine learning-based segmentation and recommendation systems, businesses can transform raw transaction data into strategic customer intelligence, driving targeted marketing campaigns, personalized customer experiences, and measurable business growth.

This project aims to bridge the gap between complex data science techniques and practical business decision-making through an intuitive, self-service analytics platform.
